we have certain features consistently have missing values in the same rows, this suggests that the data is not missing completely at random (MCAR).
For instance, missing values in features related to air pollutants like NO, NO2, and NOx may indicate a pattern based on the time of year or geographical location.
Such patterns imply that the missingness could be dependent on other observed variables, reinforcing the need for a **multivariate approach**

```python
# Calculate the correlation matrix
correlation_matrix = new_df2.drop(['City', "Date","AQI_Bucket"], axis=1).corr()
missing_data = new_df2.drop(['City', "Date","AQI_Bucket"], axis=1).isnull().sum()


# Display the correlation matrix
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()

# Look at correlations with missing data columns
missing_columns = missing_data[missing_data > 0].index
print("\nCorrelation of Features with Missing Data:")
print(correlation_matrix[missing_columns])
```

High correlations between features suggest that one feature can provide information about another, which can be used to fill its missing values.

For example, if NO2 is highly correlated with NOx, imputing missing NO2 values using information from NOx (and possibly other related variables) is likely to produce more accurate and meaningful results.

This means it makes most sense to use a multivariate approach and hence use **MICE**

# Summary
In this analysis, we addressed missing values in the dataset using the Multiple Imputation by Chained Equations (MICE) technique. To enhance the accuracy and robustness of the imputation process, we used three different estimators: Bayesian Regression, Elastic Net, and Ridge Regression. These estimators were used in an iterative process to predict and fill in missing values based on the relationships between features.

Before applying the MICE technique, we preprocessed the dataset by removing outliers. This step was crucial to ensure that the imputation models were not skewed by extreme values, thereby improving the quality and reliability of the imputations.

Overall, this approach allowed us to handle missing data effectively and prepare a cleaner dataset for subsequent analysis and modeling tasks.
